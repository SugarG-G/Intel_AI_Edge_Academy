### 1. Numpy
##### List 데이터형의 매트릭스 연산 속도

```python
#TODO 필요한 라이브러리를 import.
# time, random
from random import random
import time

#TODO 행, 열의 크기를 설정.
N=3

#TODO 행렬을 초기화합니다 (list  형).
# metA, metB, metC
matA = []
matB = []
matC = []
def make_N_by_N_list(matrix:list, n, fn):
    for _ in range(n):
        row = []
        for _ in range(n):
            row.append(random()) if fn == 1 else row.append(0)
        matrix.append(row)
    return matrix

#TODO make_N_by_N_list를 이용해서 metA, metB, metC
matA = make_N_by_N_list(matA, N, 1)
matB = make_N_by_N_list(matB, N, 1)
matC = make_N_by_N_list(matC, N, 0)

# 파이썬의 리스트를 사용하여 계산합니다.
# 시작 시간을 저장합니다.
s1 = time.time()
#TODO for 문을 사용하여 행렬 곱셈을 실행합니다.
for i in range(N):
    for j in range(N):
        for k in range(N):
            matC[i][j] += matA[i][k] * matB[k][j]
s2 = time.time()
print("list matmul time: ", s2-s1)
```

#### 차원과 관련된 함수

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 400)  # 해상도 ↑

fig, axes = plt.subplots(2, 3, figsize=(12, 6))

# (1,1): sin
ax = axes[0, 0]
y = np.sin(x)
ax.plot(x, y, label="sin", color = "red")
ax.set_title("sin")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_xticks(np.linspace(0, 2*np.pi, 9))
ax.set_yticks(np.linspace(-1, 1, 5))
ax.grid(True)
ax.legend(loc="lower right")

# (1,2): sin + cos
ax = axes[0, 1]
ax.plot(x, np.sin(x), label="sin", color = "red")
ax.plot(x, np.cos(x), label="cos", color = "blue")
ax.set_title("sin & cos")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.set_xticks(np.linspace(0, 2*np.pi, 9))
ax.set_yticks(np.linspace(-1, 1, 5))
ax.grid(True)
ax.legend(loc="lower right")

# (1,3): log(x)  -> x>0만
ax = axes[0, 2]
xpos = np.linspace(1e-3, 2*np.pi, 400)
ax.plot(xpos, np.log(xpos), label="log", color = "green")
ax.set_title("log")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.grid(True)
ax.legend(loc="lower right")

# (2,1): exp(x) -> 큰 스케일, ytick 수동 제거
ax = axes[1, 0]
ax.plot(x, np.exp(x), label="exp", color = "orange")
ax.set_title("exp")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.grid(True)
ax.legend(loc="lower right")

# (2,2): (1,2) 내용 반복
ax = axes[1, 1]
ax.plot(x, x, label="x", color = "magenta")
ax.set_title("x")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.grid(True)
ax.legend(loc="lower right")

# (2,3): x^2
ax = axes[1, 2]
ax.plot(x, x**2, label="x^2", color = "cyan")
ax.set_title("x^2")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.grid(True)
ax.legend(loc="lower right")

plt.tight_layout()
plt.show()
```
