### 1. Numpy
##### List 데이터형의 매트릭스 연산 속도

```python
#TODO 필요한 라이브러리를 import.
# time, random
from random import random
import time

#TODO 행, 열의 크기를 설정.
N=3

#TODO 행렬을 초기화합니다 (list  형).
# metA, metB, metC
matA = []
matB = []
matC = []
def make_N_by_N_list(matrix:list, n, fn):
    for _ in range(n):
        row = []
        for _ in range(n):
            row.append(random()) if fn == 1 else row.append(0)
        matrix.append(row)
    return matrix

#TODO make_N_by_N_list를 이용해서 metA, metB, metC
matA = make_N_by_N_list(matA, N, 1)
matB = make_N_by_N_list(matB, N, 1)
matC = make_N_by_N_list(matC, N, 0)

# 파이썬의 리스트를 사용하여 계산합니다.
# 시작 시간을 저장합니다.
s1 = time.time()
#TODO for 문을 사용하여 행렬 곱셈을 실행합니다.
for i in range(N):
    for j in range(N):
        for k in range(N):
            matC[i][j] += matA[i][k] * matB[k][j]
s2 = time.time()
print("list matmul time: ", s2-s1)
```

##### Q5. 2차원 배열을 90도 회전시키고, 상하 반전시키고 결과를 출력.

```python
import numpy as np
from numpy.random import rand
##### Q1. 1차원 배열을 선언하고 윈도우 크기가 3일 때 이동 평균을 계산.
arr1 = rand(20)
window = 3
result = []
#   window
# -1 | 0 | 1
# -2 | -1 | 0 | 1 | 2
hw = np.floor(window/2).astype(int)
for i in range(hw, len(arr1)-hw):
    #print(arr1[i-hw:i+hw+1])
    result.append(
        np.mean(arr1[i-hw:i+hw+1])
    )
print(result)

##### Q2. (N, N) 2차원 배열을 선언하고 지그제그로 나타내고, 이 2차원 배열을 1차원으로 변환.
###### Q2.0 (N, N) 2차원 배열을 선언하고 순서대로 나타내고, 이 2차원 배열을 1차원으로 변환.
N = 3#int(input())
arr2 = np.zeros((N,N))
for i in range(N):
    for j in range(N):
        arr2[i, j] = N*i + j
print(arr2)

arr21 = np.zeros((N,N))
for i in range(N):
    if i%2 == 0:
        for j in range(N):
            arr21[i, j] = N*i + j
    else:
        for j in range(N):
            arr21[i, N-j-1] = N*i + j
print(arr21)

##### Q3. 1차원 배열의 각 요소의 크기의 순서를 나타내시오.
arr3 = rand(20)
print(np.sort(arr3))

##### Q4. 1차원 배열에서 연속된 부분 수열의 합이 최대가 되는 구간을 찾고, 반환.
arr4 = rand(20)
result = []
hw = np.floor(window/2).astype(int)
for i in range(hw, len(arr4)-hw):
    #print(arr1[i-hw:i+hw+1])
    result.append(
        (np.sum(arr4[i-hw:i+hw+1]))
    )
pos = np.argmax(result, axis=0)

print(pos)
##### Q5. 2차원 배열을 90도 회전시키고, 상하 반전시키고 결과를 출력.
N = 3
arr5 = rand(N, N)
print(arr5)
print(arr5.T)
print(np.flip(arr5.T, axis=0))
```
